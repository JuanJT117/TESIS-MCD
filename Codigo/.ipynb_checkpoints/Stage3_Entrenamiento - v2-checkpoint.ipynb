```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV, RandomizedSearchCV , KFold
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor, ExtraTreesRegressor
from sklearn.svm import SVR
from sklearn.multioutput import MultiOutputRegressor
from sklearn.preprocessing import StandardScaler

import joblib

# Metricas

from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
from sklearn.model_selection import cross_val_score
```

```python
df = pd.read_csv('DATOS/Conjunto_Datatraining.csv')
print(df.head())
```

```python
# 1. Configuración de Columnas
df_ml = df.copy()
ab2_cols = [f"AB2_{i}" for i in range(1, 31)]
rhoa_cols = [f"Rhoa_{i}" for i in range(1, 31)]
indicator_cols = [f"is_observed_{i}" for i in range(1, 31)]

# Definimos X e y base
X_cols = ab2_cols + rhoa_cols + indicator_cols
y_cols = rhoa_cols

# 2. Split (Entrenamiento / Prueba)
# Nota: Si tienes columna 'id_siva' o similar, considera GroupShuffleSplit en el futuro.
df_train, df_test = train_test_split(df_ml, test_size=0.2, random_state=42)

# 3. Preparación de TARGETS (y) en espacio LOGARÍTMICO
# Aplicamos log10 a las salidas. Sumamos un epsilon minúsculo por seguridad física, 
# aunque tus datos originales no deberían tener ceros.
y_train_log = np.log10(df_train[y_cols].copy() + 1e-9)
y_test_log = np.log10(df_test[y_cols].copy() + 1e-9)

# Guardamos copias lineales para evaluación final
y_test_real = df_test[y_cols].copy() 

# 4. Enmascaramiento (Data Masking) en espacio LINEAL primero
# Inicializamos indicadores
for df_set in [df_train, df_test]:
    for col in indicator_cols:
        df_set[col] = 1.0

# Aplicamos huecos aleatorios
np.random.seed(42)
mask_prob = 0.2
sentinel_value_linear = 0.0 # Valor temporal en lineal

for df_set in [df_train, df_test]:
    for i in range(1, 31):
        col = f"Rhoa_{i}"
        ind_col = f"is_observed_{i}"
        mask = np.random.rand(len(df_set)) < mask_prob
        
        # Aplicar daño
        df_set.loc[mask, col] = sentinel_value_linear
        df_set.loc[mask, ind_col] = 0.0

# 5. Transformación LOGARÍTMICA de los INPUTS (X)
# Aquí está el truco: Manejar los ceros del enmascaramiento.

def log_transform_inputs(df_input, rhoa_columns):
    df_log = df_input.copy()
    
    # Solo transformamos las columnas de resistividad
    for col in rhoa_columns:
        # Extraemos la serie
        series = df_log[col]
        
        # Caso 1: Datos válidos (mayores a 0) -> Log10
        valid_mask = series > 0
        df_log.loc[valid_mask, col] = np.log10(series[valid_mask])
        
        # Caso 2: Datos enmascarados (igual a 0) -> Valor Centinela (-10.0)
        # -10 significa 1e-10 Ohm-m, físicamente imposible, sirve de señal clara al RF.
        df_log.loc[~valid_mask, col] = -10.0
        
    return df_log[X_cols] # Retornamos solo las columnas X

X_train_log = log_transform_inputs(df_train, rhoa_cols)
X_test_log = log_transform_inputs(df_test, rhoa_cols)

print("Transformación Logarítmica completada.")
print(f"Rango de valores en y_train_log: {y_train_log.min().min():.2f} a {y_train_log.max().max():.2f}")
print("Valor usado para huecos en inputs (X): -10.0")

```

# Random Forest


```python
# Definir modelo base y wrapper
rf_base = RandomForestRegressor(random_state=42)
model_wrapper = MultiOutputRegressor(rf_base)

# Grid de Hiperparámetros (Estrategia Global)
# Ajustado para ser riguroso pero no eterno
param_grid_global = {
    "estimator__n_estimators": [100, 200],
    "estimator__max_depth": [10, 15, 20, 30, None],     # None deja crecer los árboles (cuidado overfitting)
    "estimator__min_samples_split": [2, 5],
    "estimator__min_samples_leaf": [1, 2],      # leaf=2 ayuda a suavizar ruido
    "estimator__max_features": ['sqrt', None]   # sqrt es excelente para Random Forest
}

# Configurar GridSearchCV
grid_search = GridSearchCV(
    estimator=model_wrapper,
    param_grid=param_grid_global,
    cv=3,
    scoring="r2", # Optimizará R2 en espacio logarítmico (que es bueno para la tendencia)
    n_jobs=-1,
    verbose=1
)

# Entrenar
print("Iniciando entrenamiento con Inputs/Outputs Logarítmicos...")
grid_search.fit(X_train_log, y_train_log)

# Asignar mejor modelo a variable estándar
rf_multi = grid_search.best_estimator_

# Guardar
joblib.dump(rf_multi, "rf_log_model.joblib")
print(f"Mejores parámetros: {grid_search.best_params_}")
joblib.dump(X_cols, "random_forest_feature_columns_v2.joblib")

print("Modelo entrenado con GridSearchCV. Mejores parámetros disponibles en rf_multi.estimators_[0].best_params_.")
```

```python
# ==============================================================================
# EVALUACIÓN ESTANDARIZADA - RANDOM FOREST (RF)
# ==============================================================================
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, mean_squared_log_error
from scipy.stats import skew, kurtosis, shapiro
import matplotlib.pyplot as plt
import numpy as np

# A. PREDICCIÓN Y TRANSFORMACIÓN INVERSA
# ------------------------------------------------------------------------------
print("\n>>> GENERANDO PREDICCIONES CON RANDOM FOREST (RF)...")
y_pred_log = rf_multi.predict(X_test_log)
y_pred_real = 10 ** y_pred_log  # Inversa: Log10 -> Ohm.m

# Ground Truth Físico
y_test_real_array = y_test_real.to_numpy() if hasattr(y_test_real, 'to_numpy') else y_test_real

# B. CÁLCULO DE MÉTRICAS GLOBALES
# ------------------------------------------------------------------------------
# 1. R2 Global
r2_global = r2_score(y_test_real_array, y_pred_real, multioutput='uniform_average')
# 2. MSE & RMSE
mse_global = mean_squared_error(y_test_real_array, y_pred_real)
rmse_global = np.sqrt(mse_global)
# 3. MAE
mae_global = mean_absolute_error(y_test_real_array, y_pred_real)
# 4. MSLE (Mean Squared Logarithmic Error) - NUEVO
# Nota: Sumamos epsilon si hubiera ceros, pero en resistividad física no debe haberlos.
try:
    msle_global = mean_squared_log_error(y_test_real_array, y_pred_real)
except ValueError:
    msle_global = np.nan
    print("Advertencia: Valores negativos detectados, no se pudo calcular MSLE.")

# 5. MAPE (Métrica Personalizada)
def calculate_mape(y_true, y_pred):
    mask = y_true > 1e-9 
    if mask.sum() > 0:
        return np.mean(np.abs((y_true[mask] - y_pred[mask]) / y_true[mask])) * 100
    return np.nan
mape_global = calculate_mape(y_test_real_array, y_pred_real)

# C. REPORTE TÉCNICO
# ------------------------------------------------------------------------------
print("-" * 40)
print(f"REPORTE DE DESEMPEÑO: RANDOM FOREST")
print("-" * 40)
print(f"R² Global (Varianza Explicada): {r2_global:.4f}")
print(f"RMSE (Error Global):            {rmse_global:.2f} Ohm.m")
print(f"MAE (Error Absoluto):           {mae_global:.2f} Ohm.m")
print(f"MAPE (Error Porcentual):        {mape_global:.2f} %")
print(f"MSLE (Error Logarítmico):       {msle_global:.4f}")
print("-" * 40)

# D. ANÁLISIS POR PROFUNDIDAD (PRIMERAS 5 CAPAS)
# ------------------------------------------------------------------------------
print("Detalle por Capa (Primeras 5 profundidades):")
residuals = y_test_real_array - y_pred_real
for i in range(30):
    # Métricas locales
    r2_col = r2_score(y_test_real_array[:, i], y_pred_real[:, i])
    rmse_col = np.sqrt(mean_squared_error(y_test_real_array[:, i], y_pred_real[:, i]))
    
    # Estadística de Residuos
    res_col = residuals[:, i]
    skew_val = skew(res_col)
    kurt_val = kurtosis(res_col)
    try:
        stat, p_val = shapiro(res_col) if len(res_col) < 5000 else (0, 0)
    except: p_val = 0
    
    print(f"Capa {i+1} | R²: {r2_col:.3f} | RMSE: {rmse_col:.1f} | Asimetría: {skew_val:.2f} | Curtosis: {kurt_val:.2f}")

# E. HIPERPARÁMETROS
# ------------------------------------------------------------------------------
print("-" * 40)
if hasattr(grid_search, 'best_params_'):
    print(f"Mejores Hiperparámetros (RF):\n{grid_search.best_params_}")
else:
    print("Modelo cargado o sin GridSearch activo.")

# F. GRÁFICOS DE RESIDUOS (ESTILO RF: TEAL)
# ------------------------------------------------------------------------------
for i in range(30):
    plt.figure(figsize=(8, 4))
    plt.scatter(y_pred_real[:, i], residuals[:, i], 
                c='teal', alpha=0.5, edgecolors='k', s=30, label='Residuals')
    plt.axhline(y=0, color='r', linestyle='--', linewidth=2, label='Zero Error')
    plt.xlabel(f'Predicción RF (Rhoa Capa {i+1}) [$\Omega\cdot m$]')
    plt.ylabel('Residuo ($Obs - Pred$)')
    plt.title(f'Análisis de Residuos RF - Capa {i+1}')
    plt.grid(True, alpha=0.5)
    plt.legend()
    plt.savefig(f'rf_residual_plot_col_{i+1}.png', dpi=300, bbox_inches='tight')
    plt.show()
    plt.close()
```

# Support Vector Machine (SVM)

```python
from sklearn.svm import SVR
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

# 1. Definir la tubería (Pipeline) de procesamiento
# Primero Escala -> Luego aplica SVM
# Esto asegura que el escalado se ajuste SOLO con el train set en cada validación cruzada (evita data leakage).
svm_pipeline = Pipeline([
    ('scaler', StandardScaler()), 
    ('svm', SVR(max_iter=50000)) # Límite de iteraciones por seguridad
])

# 2. Envolver en MultiOutputRegressor para predecir toda la curva
# El "estimator" ahora es el pipeline completo
svm_wrapper = MultiOutputRegressor(svm_pipeline)

# 3. Grid de Hiperparámetros Expandido (Nomenclatura anidada)
# Sintaxis: estimator__[nombre_paso_pipeline]__[parametro]
param_grid_svm_global = {
    # C (Regularización): Controla el equilibrio entre error y simplicidad.
    # C alto (1000): Intenta clasificar todo perfecto (riesgo de overfitting).
    # C bajo (1): Prefiere una frontera de decisión suave (underfitting).
    "estimator__svm__C": [1, 10, 100, 500, 1000],

    # Epsilon (Insensibilidad): Margen de tolerancia donde el error no cuenta.
    # En escala logarítmica: 0.1 significa tolerar un error del ~25%. 
    # 0.01 es muy estricto (bueno para datos limpios sintéticos).
    "estimator__svm__epsilon": [0.01, 0.05, 0.1, 0.2, 0.5],

    # Kernel: La función matemática que transforma el espacio.
    # 'rbf' (Radial Basis Function) es ideal para curvas suaves como SEV.
    # 'poly' puede servir, pero a veces oscila mucho.
    "estimator__svm__kernel": ['rbf', 'linear'], 

    # Gamma (Coeficiente del Kernel): Define la "influencia" de un solo ejemplo.
    # 'scale': 1 / (n_features * X.var()) -> Valor seguro por defecto.
    # 0.1, 0.01: Valores manuales para ajuste fino.
    "estimator__svm__gamma": ['scale', 0.1, 0.01]
}

# 4. Configuración de GridSearchCV Global
# n_jobs=8 (o -1) para usar todos los núcleos. SVM es pesado computacionalmente.
grid_search_svm = GridSearchCV(
    estimator=svm_wrapper,
    param_grid=param_grid_svm_global,
    cv=3,
    scoring="r2", # R2 en espacio logarítmico
    n_jobs=-1,    # Usar todos los núcleos disponibles
    verbose=2
)

# 5. Entrenar
print("Iniciando entrenamiento SVM Global con Datos Logarítmicos y Escalado...")
# Usamos X_train_log y y_train_log (generados en el paso de Random Forest)
grid_search_svm.fit(X_train_log, y_train_log)

# 6. Resultados y Persistencia
print("-" * 30)
print(f"Mejores parámetros SVM: {grid_search_svm.best_params_}")

# Extraer el mejor modelo y asignarlo a la variable estándar
svm_multi = grid_search_svm.best_estimator_

# Guardar
joblib.dump(svm_multi, "svm_multioutput_opt_log.joblib")
print("Modelo SVM optimizado guardado exitosamente.")

# --- EVALUACIÓN RÁPIDA (Opcional, para verificar) ---
# Predecir en test
y_pred_log_svm = svm_multi.predict(X_test_log)
y_pred_real_svm = 10 ** y_pred_log_svm # Inversa del log

# Calcular métrica rápida
from sklearn.metrics import mean_squared_error
rmse_svm = np.sqrt(mean_squared_error(y_test_real.to_numpy(), y_pred_real_svm))
print(f"RMSE Global SVM (Ohm-m): {rmse_svm:.2f}")
```

```python
# ==============================================================================
# EVALUACIÓN ESTANDARIZADA - SUPPORT VECTOR MACHINE (SVM)
# ==============================================================================
# Nota: Se asume que las librerías ya se cargaron en el bloque anterior

# A. PREDICCIÓN Y TRANSFORMACIÓN INVERSA
# ------------------------------------------------------------------------------
print("\n>>> GENERANDO PREDICCIONES CON SVM...")
y_pred_log = svm_multi.predict(X_test_log)
y_pred_real = 10 ** y_pred_log 
y_test_real_array = y_test_real.to_numpy() if hasattr(y_test_real, 'to_numpy') else y_test_real

# B. CÁLCULO DE MÉTRICAS GLOBALES
# ------------------------------------------------------------------------------
r2_global = r2_score(y_test_real_array, y_pred_real, multioutput='uniform_average')
mse_global = mean_squared_error(y_test_real_array, y_pred_real)
rmse_global = np.sqrt(mse_global)
mae_global = mean_absolute_error(y_test_real_array, y_pred_real)
try:
    msle_global = mean_squared_log_error(y_test_real_array, y_pred_real)
except ValueError:
    msle_global = np.nan

mape_global = calculate_mape(y_test_real_array, y_pred_real)

# C. REPORTE TÉCNICO
# ------------------------------------------------------------------------------
print("-" * 40)
print(f"REPORTE DE DESEMPEÑO: SUPPORT VECTOR MACHINE (SVM)")
print("-" * 40)
print(f"R² Global (Varianza Explicada): {r2_global:.4f}")
print(f"RMSE (Error Global):            {rmse_global:.2f} Ohm.m")
print(f"MAE (Error Absoluto):           {mae_global:.2f} Ohm.m")
print(f"MAPE (Error Porcentual):        {mape_global:.2f} %")
print(f"MSLE (Error Logarítmico):       {msle_global:.4f}")
print("-" * 40)

# D. ANÁLISIS POR PROFUNDIDAD
# ------------------------------------------------------------------------------
print("Detalle por Capa (Primeras 5 profundidades):")
residuals = y_test_real_array - y_pred_real
for i in range(30):
    r2_col = r2_score(y_test_real_array[:, i], y_pred_real[:, i])
    rmse_col = np.sqrt(mean_squared_error(y_test_real_array[:, i], y_pred_real[:, i]))
    res_col = residuals[:, i]
    skew_val = skew(res_col)
    kurt_val = kurtosis(res_col)
    print(f"Capa {i+1} | R²: {r2_col:.3f} | RMSE: {rmse_col:.1f} | Asimetría: {skew_val:.2f} | Curtosis: {kurt_val:.2f}")

# E. HIPERPARÁMETROS
# ------------------------------------------------------------------------------
print("-" * 40)
# Lógica específica para SVM Pipeline
if 'grid_search_svm' in locals() and hasattr(grid_search_svm, 'best_params_'):
     print(f"Mejores Hiperparámetros (SVM):\n{grid_search_svm.best_params_}")
elif hasattr(svm_multi, 'estimators_'):
    try:
        # Extraer del primer estimador del pipeline
        est = svm_multi.estimators_[0].named_steps['svm']
        print(f"Params SVM (Cargados): C={est.C}, Kernel={est.kernel}, Gamma={est.gamma}")
    except: print("No se pudieron extraer params detallados.")

# F. GRÁFICOS DE RESIDUOS (ESTILO SVM: ORANGE)
# ------------------------------------------------------------------------------
for i in range(30):
    plt.figure(figsize=(8, 4))
    plt.scatter(y_pred_real[:, i], residuals[:, i], 
                c='darkorange', alpha=0.6, edgecolors='k', s=25, label='Residuals')
    plt.axhline(y=0, color='r', linestyle='--', linewidth=2, label='Zero Error')
    plt.xlabel(f'Predicción SVM (Rhoa Capa {i+1}) [$\Omega\cdot m$]')
    plt.ylabel('Residuo ($Obs - Pred$)')
    plt.title(f'Análisis de Residuos SVM - Capa {i+1}')
    plt.grid(True, alpha=0.5)
    plt.legend()
    plt.savefig(f'svm_residual_plot_col_{i+1}.png', dpi=300, bbox_inches='tight')
    plt.show()
    plt.close()
```

# Gradient Boosting Regressor (GBR)

```python
from sklearn.ensemble import GradientBoostingRegressor

# 1. Definir Modelo Base y Wrapper
# random_state=42 asegura reproducibilidad científica
gbr = GradientBoostingRegressor(random_state=42)
gbr_wrapper = MultiOutputRegressor(gbr)

# 2. Grid de Hiperparámetros Expandido (Nomenclatura Global)
# Nota: GBR es sensible a la relación entre learning_rate y n_estimators.
# Tasas bajas (0.01) requieren más árboles (500) pero generalizan mejor.
param_grid_gbr_global = {
    "estimator__n_estimators": [100, 200, 300, 500],
    
    # La velocidad a la que el modelo corrige errores. 
    # 0.05 suele ser un punto dulce para datos con ruido moderado.
    "estimator__learning_rate": [0.01, 0.05, 0.07, 0.1],
    
    # Profundidad: GBR suele funcionar bien con árboles "chatos" (3-5).
    # Profundidades > 5 suelen memorizar ruido en datasets pequeños.
    "estimator__max_depth": [2, 3, 4, 5, 7, 10],
    
    # Stochastic Gradient Boosting: Usar < 1.0 reduce varianza
    "estimator__subsample": [0.8, 0.9, 1.0],
    
    # Regularización extra para evitar "perseguir" outliers
    "estimator__min_samples_leaf": [1, 2]
}

# 3. Configuración de GridSearchCV Global
# Usamos n_jobs=-1, pero ojo: GBR es secuencial por naturaleza, 
# la paralelización ocurre en la búsqueda del Grid, no dentro del ajuste de un solo modelo.
gbr_multi = GridSearchCV(
    estimator=gbr_wrapper,
    param_grid=param_grid_gbr_global,
    cv=3,
    scoring="r2", # R2 en espacio Logarítmico
    n_jobs=-1,
    verbose=1
)

# 4. Entrenamiento
print("Iniciando entrenamiento GBR Global con Datos Logarítmicos...")
# Importante: Usamos X_train_log y y_train_log (definidos en la etapa RF)
gbr_multi.fit(X_train_log, y_train_log)

# 5. Guardado
# Extraemos el mejor estimador del Grid
best_gbr_model = gbr_multi.best_estimator_

joblib.dump(best_gbr_model, "gbr_multioutput_opt_log.joblib")
joblib.dump(X_cols, "gbr_feature_columns_v2.joblib")

print("-" * 30)
print(f"Mejores parámetros GBR: {gbr_multi.best_params_}")
```

```python
# ==============================================================================
# EVALUACIÓN ESTANDARIZADA - GRADIENT BOOSTING (GBR)
# ==============================================================================
# Nota: Se asume que las librerías ya se cargaron en el bloque anterior

# A. PREDICCIÓN Y TRANSFORMACIÓN INVERSA
# ------------------------------------------------------------------------------
print("\n>>> GENERANDO PREDICCIONES CON GRADIENT BOOSTING (GBR)...")
y_pred_log = gbr_multi.predict(X_test_log)
y_pred_real = 10 ** y_pred_log 
y_test_real_array = y_test_real.to_numpy() if hasattr(y_test_real, 'to_numpy') else y_test_real

# B. CÁLCULO DE MÉTRICAS GLOBALES
# ------------------------------------------------------------------------------
r2_global = r2_score(y_test_real_array, y_pred_real, multioutput='uniform_average')
mse_global = mean_squared_error(y_test_real_array, y_pred_real)
rmse_global = np.sqrt(mse_global)
mae_global = mean_absolute_error(y_test_real_array, y_pred_real)
try:
    msle_global = mean_squared_log_error(y_test_real_array, y_pred_real)
except ValueError:
    msle_global = np.nan

mape_global = calculate_mape(y_test_real_array, y_pred_real)

# C. REPORTE TÉCNICO
# ------------------------------------------------------------------------------
print("-" * 40)
print(f"REPORTE DE DESEMPEÑO: GRADIENT BOOSTING (GBR)")
print("-" * 40)
print(f"R² Global (Varianza Explicada): {r2_global:.4f}")
print(f"RMSE (Error Global):            {rmse_global:.2f} Ohm.m")
print(f"MAE (Error Absoluto):           {mae_global:.2f} Ohm.m")
print(f"MAPE (Error Porcentual):        {mape_global:.2f} %")
print(f"MSLE (Error Logarítmico):       {msle_global:.4f}")
print("-" * 40)

# D. ANÁLISIS POR PROFUNDIDAD
# ------------------------------------------------------------------------------
print("Detalle por Capa (Primeras 5 profundidades):")
residuals = y_test_real_array - y_pred_real
for i in range(30):
    r2_col = r2_score(y_test_real_array[:, i], y_pred_real[:, i])
    rmse_col = np.sqrt(mean_squared_error(y_test_real_array[:, i], y_pred_real[:, i]))
    res_col = residuals[:, i]
    skew_val = skew(res_col)
    kurt_val = kurtosis(res_col)
    print(f"Capa {i+1} | R²: {r2_col:.3f} | RMSE: {rmse_col:.1f} | Asimetría: {skew_val:.2f} | Curtosis: {kurt_val:.2f}")

# E. HIPERPARÁMETROS
# ------------------------------------------------------------------------------
print("-" * 40)
if 'gbr_multi' in locals() and hasattr(gbr_multi, 'best_params_'):
    print(f"Mejores Hiperparámetros (GBR):\n{gbr_multi.best_params_}")
elif hasattr(gbr_multi, 'estimators_'):
    try:
        est = gbr_multi.estimators_[0]
        # Filtramos los importantes
        p = {k:v for k,v in est.get_params().items() if k in ['n_estimators', 'learning_rate', 'max_depth']}
        print(f"Params GBR (Cargados): {p}")
    except: print("No se pudieron extraer params detallados.")

# F. GRÁFICOS DE RESIDUOS (ESTILO GBR: ROYAL BLUE)
# ------------------------------------------------------------------------------
for i in range(30):
    plt.figure(figsize=(8, 4))
    plt.scatter(y_pred_real[:, i], residuals[:, i], 
                c='royalblue', alpha=0.6, edgecolors='white', s=30, label='Residuals')
    plt.axhline(y=0, color='k', linestyle='--', linewidth=2, label='Zero Error')
    plt.xlabel(f'Predicción GBR (Rhoa Capa {i+1}) [$\Omega\cdot m$]')
    plt.ylabel('Residuo ($Obs - Pred$)')
    plt.title(f'Análisis de Residuos GBR - Capa {i+1}')
    plt.grid(True, alpha=0.5)
    plt.legend()
    plt.savefig(f'gbr_residual_plot_col_{i+1}.png', dpi=300, bbox_inches='tight')
    plt.show()
    plt.close()
```
